```python
### ====== OPTUNA-BASED BAYESIAN OPTIMIZATION FOR FUNCTION 4 ======
# Import the OptunaBayesianOptimizer class
import sys
sys.path.insert(0, './bayesian_optimization_challenge-md')
from bo_optuna import OptunaBayesianOptimizer

# Create optimizer instance with initial Function 4 data
print("=" * 60)
print("OPTUNA-BASED BAYESIAN OPTIMIZATION FOR FUNCTION 4")
print("=" * 60)

X_func4_initial = week_0_func4_inputs
y_func4_initial = week_0_func4_outputs
bounds_func4 = [(0, 1)] * X_func4_initial.shape[1]

optimizer_func4 = OptunaBayesianOptimizer(
    X_initial=X_func4_initial,
    y_initial=y_func4_initial,
    bounds=bounds_func4,
    optimize_hp=True,  # Enable hyperparameter tuning
    random_state=42,
    acquisition="ucb"
)

print(f"\nInitial training data shape: X={optimizer_func4.X_train.shape}, y={optimizer_func4.y_train.shape}")
print(f"Initial best observation: {optimizer_func4.get_best_observation()[1]:.6e}")
```

```python
# Run Optuna-based BO for 8 weeks (8 iterations)
print("\nRunning Optuna-based BO for 8 iterations...")
print("-" * 60)

# Get all weekly data
weekly_data = [
    (week_1_inputs[3], week_1_outputs[3]),
    (week_2_inputs[3], week_2_outputs[3]),
    (week_3_inputs[3], week_3_outputs[3]),
    (week_4_inputs[3], week_4_outputs[3]),
    (week_5_inputs[3], week_5_outputs[3]),
    (week_6_inputs[3], week_6_outputs[3]),
    (week_7_inputs[3], week_7_outputs[3]),
    (week_8_inputs[3], week_8_outputs[3])
]

optuna_proposals_func4 = []
manual_best_func4 = week_0_func4_outputs.max()
optuna_best_func4 = y_func4_initial.max()

for week, (x_actual, y_actual) in enumerate(weekly_data, start=1):
    print(f"\nWeek {week}:")
    print(f"  Actual observation: y = {y_actual:.6e}")
    
    # Get Optuna proposal
    proposals = optimizer_func4.optimize(
        n_iterations=1,
        optimize_hp_every=1 if week % 2 == 0 else 0,  # Tune HP every other week
        optimize_hp_n_trials=30,
        acq_n_trials=100,
        verbose=True
    )
    
    x_proposed = proposals[0]
    optuna_proposals_func4.append(x_proposed)
    
    # Update optimizer with actual observation
    optimizer_func4.update(x_actual, y_actual)
    
    # Track best values
    manual_best_func4 = max(manual_best_func4, y_actual)
    optuna_best_func4 = max(optuna_best_func4, y_actual)
    
    print(f"  Best so far (Optuna): {optuna_best_func4:.6e}")

print("\n" + "=" * 60)
print("OPTUNA-BASED BO COMPLETED FOR FUNCTION 4")
print("=" * 60)
```
