```python
import os
import numpy as np

def ensure_dir(path: str):
    os.makedirs(path, exist_ok=True)

def as_2d_row(x):
    x = np.asarray(x)
    if x.ndim == 1:
        return x.reshape(1, -1)
    if x.ndim == 2 and x.shape[0] == 1:
        return x
    raise ValueError(f"Expected a 1D vector or a (1, d) row vector. Got shape {x.shape}.")

def load_week0(func_number: int):
    inputs_path = f'./initial_data/function_{func_number}/initial_inputs.npy'
    outputs_path = f'./initial_data/function_{func_number}/initial_outputs.npy'
    return np.load(inputs_path), np.load(outputs_path)

def update_one_function(func_number: int, new_input, new_output):
    week0_inputs, week0_outputs = load_week0(func_number)

    new_input_row = as_2d_row(new_input)

    expected_d = week0_inputs.shape[1]
    actual_d = new_input_row.shape[1]

    if actual_d != expected_d:
        raise ValueError(
            f"Function {func_number}: Week 0 expects d={expected_d}, "
            f"but new input has d={actual_d}. New input: {np.asarray(new_input)}"
        )

    combined_inputs = np.vstack([week0_inputs, new_input_row])
    combined_outputs = np.append(week0_outputs, new_output)

    return combined_inputs, combined_outputs

import os
import numpy as np

def save_combined_week(func_number: int, combined_inputs, combined_outputs, out_root: str):
    out_dir = os.path.join(out_root, f'function_{func_number}')
    ensure_dir(out_dir)

    # Save NumPy binaries
    # np.save(os.path.join(out_dir, 'inputs.npy'), combined_inputs)
    # np.save(os.path.join(out_dir, 'outputs.npy'), combined_outputs)

    # -------- Save CSV --------
    
    # Inputs CSV (each column = feature)
    inputs_csv_path = os.path.join(out_dir, 'inputs.csv')
    np.savetxt(
        inputs_csv_path,
        combined_inputs,
        delimiter=',',
        fmt='%.6f'
    )

    # Outputs CSV (single column)
    outputs_csv_path = os.path.join(out_dir, 'outputs.csv')
    np.savetxt(
        outputs_csv_path,
        combined_outputs.reshape(-1, 1),
        delimiter=',',
        fmt='%.6f'
    )

    print(f"Function {func_number} saved:")
    print(f"  {inputs_csv_path}")
    print(f"  {outputs_csv_path}")

def run_week_update(func_numbers, week_inputs, week_outputs, out_root='./combined_week_1', strict=True):
    """
    func_numbers: list of function numbers you want to update, e.g. [1,2,3,4,5,6]
    week_inputs:  list of new x's, same length as func_numbers
    week_outputs: list of new y's, same length as func_numbers
    strict: if False, skips missing folders / dimension mismatches and continues
    """
    if not (len(func_numbers) == len(week_inputs) == len(week_outputs)):
        raise ValueError("func_numbers, week_inputs, week_outputs must have the same length.")

    combined_data = {}

    print("\n=== RUNNING WEEK UPDATE ===")
    for fn, x, y in zip(func_numbers, week_inputs, week_outputs):
        try:
            combined_inputs, combined_outputs = update_one_function(fn, x, y)
            save_combined_week(fn, combined_inputs, combined_outputs, out_root=out_root)

            combined_data[f'function_{fn}'] = {
                'inputs': combined_inputs,
                'outputs': combined_outputs
            }

            print(f"Function {fn}: OK | new x={combined_inputs[-1]} | new y={combined_outputs[-1]}")
        except Exception as e:
            print(f"Function {fn}: SKIP/FAIL -> {e}")
            if strict:
                raise

    print(f"\nSaved combined arrays under: {out_root}/function_k/")
    return combined_data
```

```python
run_week_update(func_numbers=range(1, 9), week_inputs=week_1_inputs, week_outputs=week_1_outputs, out_root='./combined_week_1', strict=False)
```
